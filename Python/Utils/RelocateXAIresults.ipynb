```python
import pandas as pd
from pathlib import Path
import shutil

root = r"C:\Users\kyohe\Aerial_Photo_Classifier\20251209Data\Result_XAI"

result_csv = r"C:\Users\kyohe\Aerial_Photo_Classifier\20251209Data\Results\20251212_1758\20251212_1758_forward_ep150.csv"

# ====== 設定 ======
csv_path = Path(result_csv)      # 分類結果 CSV
images_dir = Path(root + "\\All")      # 画像格納ルート
output_dir = Path(root)   # 分類後の出力ルート

# 移動先フォルダ名
folders = ["TP", "FP", "TN", "FN"]



# ====== フォルダ準備 ======
for f in folders:
    (output_dir / f).mkdir(parents=True, exist_ok=True)

# ====== CSV 読込 ======
df = pd.read_csv(csv_path)

# ====== 判定関数 ======
def get_category(row):
    true = row["TrueLabel"]
    pred = row["PredictedLabel"]

    true = 1 if row["TrueLabel"] == "Damaged" else 0
    pred = 1 if row["PredictedLabel"] == "Damaged" else 0


    # CASE: CLASS が 1 / 0 の場合
    # TrueLabel と PredictedLabel が文字列なら True/Pred に合わせて変える
    # 例: Damaged / NotDamaged → 1 / 0 に変換済みなら不要

    if true == pred:
        # 正解
        return "TP" if pred == 1 else "TN"
    else:
        # 誤分類
        return "FP" if pred == 1 else "FN"


# ====== 画像移動 ======
for _, row in df.iterrows():
    fname = row["filename"]
    category = get_category(row)

    src = images_dir / fname

    if src.exists():
        dst = output_dir / category / fname
        shutil.move(str(src), str(dst))
        print(f"Moved {fname} → {category}")
    else:
        print(f"WARNING: {fname} not found in {images_dir}")
```
